### Homework #4: Kernel Density Estimation

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# PROBLEM 1c
data = np.array([1, 1.75, 2, 3, 3.25, 4])

h = 1.84 * data.std() * len(data) ** (-0.2)

print(f'Plug-in Bandwidth Estimated is: {h:2f}')
```

```python
X = np.array([1, 1.75, 2, 3, 3.25, 4])

h = 0.2
n = len(X)

x_grid = np.linspace(X.min() - h, X.max() + h, 30)

I = np.abs(x_grid[:, None] - X[None, :]) < h

f_hat = np.sum(I, axis=1) / (2 * n * h)

plt.step(x_grid, f_hat, where="mid")
plt.xlabel("x")
plt.ylabel("Density")
plt.title("Uniform KDE, h = 0.2")
plt.show()
```
